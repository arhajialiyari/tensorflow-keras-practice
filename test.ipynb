```python
import numpy as np
import matplotlib.pyplot as plt

# تعداد نمونه‌ها
n_samples = 1000
np.random.seed(0)
X = np.random.uniform(-1, 1, (n_samples, 2))

# تعریف لوزی: |x| + |y| < 0.7 (یک لوزی مرکزی)
y = (np.abs(X[:, 0]) + np.abs(X[:, 1]) < 0.7).astype(int)

# نمایش نمونه‌ها
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='bwr', alpha=0.5)
plt.title("Points inside and outside a diamond")
plt.xlabel("x1")
plt.ylabel("x2")
plt.show()
```
